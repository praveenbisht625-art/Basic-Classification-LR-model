# 🌸 Iris Flower Classification

## 1. Introduction
This is my first Machine Learning classification project.
The goal is to classify an iris flower into one of three species
using its flower measurements.

## 2. Dataset

The dataset contains measurements of iris flowers.

### Features:
- Sepal Length
- Sepal Width
- Petal Length
- Petal Width

### Target:
- Iris Setosa
- Iris Versicolor
- Iris Virginica


```python
import pandas as pd            # importing libraries
from sklearn.tree import DecisionTreeClassifier
```

```python
df = pd.read_csv("Iris/Iris.csv")      # Loading CSV
df
```

## Exploring The Data

# Preparing features and labels

```python
x = df[["SepalLengthCm", "SepalWidthCm", "PetalLengthCm", "PetalWidthCm"]]
y = df["Species"]
```

```python
x.head()
```

## Training the Model

```python
clf = DecisionTreeClassifier()
clf.fit(x,y)
```

```python
new_flower = pd.DataFrame(
    [[5.1, 3.5, 1.4, 0.2],[5.9, 3.0, 4.2, 1.5]],
    columns=["SepalLengthCm", "SepalWidthCm", "PetalLengthCm", "PetalWidthCm"]
)
prediction = clf.predict(new_flower)

print(f"The flower is predicted as: {prediction[0]}")
print(f"The flower is predicted as: {prediction[1]}")
```

## Decision Tree Visualization

```python
from sklearn.tree import plot_tree
import matplotlib.pyplot as plt

plt.figure(figsize=(12, 8))
plot_tree(clf, feature_names = x.columns, class_names = ["Iris-setosa","Iris-versicolor","Iris-virginica"], filled=True)

plt.tight_layout()
plt.show()
```
