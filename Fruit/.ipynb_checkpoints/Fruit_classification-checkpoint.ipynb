```python
from sklearn.tree import DecisionTreeClassifier
```

```python
# Features: [weight (grams), color (0 = red, 1 = orange)]
features = [
    [150, 0],  # Apple
    [170, 0],  # Apple
    [130, 1],  # Orange
    [120, 1],  # Orange
]
labels = ["apple", "apple", "orange", "orange"]

clf = DecisionTreeClassifier()
clf = clf.fit(features, labels)

a = clf.predict([[140, 1]])  # Light and orange? Probably an orange!
print(f" The fruit is probably {a[0]} ")

a = clf.predict([[160, 0]]) 
print(f" The fruit is probably {a[0]} ")

a = clf.predict([[115, 0]])
print(f" The fruit is probably {a[0]} ")

a = clf.predict([[170, 1]])
print(f" The fruit is probably {a[0]} ")
```

```python
from sklearn.tree import plot_tree
import matplotlib.pyplot as plt

plt.figure(figsize=(5,6))
plot_tree(clf, feature_names=["weight", "color"], class_names=["apple", "orange"], filled=True)
plt.tight_layout()
plt.show()
```
