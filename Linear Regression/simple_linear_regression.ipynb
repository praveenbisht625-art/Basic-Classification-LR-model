# Simple Linear Regression

A Linear Regression Model Predicting Salary Based on Years of Experience.

```python
import pandas as pd
import numpy as np                    # importing librariess
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("Experience-Salary.csv")    
```

```python
df.sample(10)
```

```python
df.shape
```

```python
df.info()
```

# Plotting DATASET

```python
plt.scatter(df['exp(in months)'],df['salary(in thousands)'])
plt.xlabel('exp(in months)')
plt.ylabel('salary(in thousands)')
```

```python
X = df.iloc[:,0:1]       # Extracting features
y = df.iloc[:,-1]
```

```python
y
```

```python
from sklearn.model_selection import train_test_split               # dividing data into train and test
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size = 0.3,random_state = 42)
```

```python
X_test.head()
```

```python
y_test.head()
```

```python
from sklearn.linear_model import LinearRegression
```

```python
lr = LinearRegression()        # created a LR's object
```

```python
lr.fit(X_train,y_train)
```

```python
lr.predict(X_test.iloc[2].values.reshape(1,1))
```

# Plotting Dataset to show Best Fit line

```python
plt.scatter(df['exp(in months)'],df['salary(in thousands)'])
plt.plot(X_train,lr.predict(X_train), color = 'black')
plt.xlabel('exp(in months)')
plt.ylabel('salary(in thousands)')
```

```python
m = lr.coef_
m
```

```python
b = lr.intercept_
b
```

# Regression Metrics
There are Mean Absolute Error, Mean Squared Error, Root Mean Squared Error & R2 Score

```python
from sklearn.metrics import mean_absolute_error,mean_squared_error,r2_score
```

```python
y_pred = lr.predict(X_test)
```

```python
y_test.values
```

```python
print("Mean absoulte error is ->",mean_absolute_error(y_test,y_pred))
```

```python
print("Mean squared error is ->",mean_squared_error(y_test,y_pred))
```

```python
print("Root mean squared error is ->",np.sqrt(mean_squared_error(y_test,y_pred)))
```

```python
print("R2 score ->",r2_score(y_test,y_pred))
```
